```python
def blend_linear(frame_1, frame_2, w):
    frame = (1-w) * frame_1 + w * frame_2
    return frame

def blend_linear_kb(frame_1: torch.Tensor, frame_2: torch.Tensor, w) -> torch.Tensor:
    """
    frame_*: [N, B, 3]
    w: scalar or [N], [N,1], [N,B], [N,B,1], [N,1,1]
    """
    assert frame_1.shape == frame_2.shape and frame_1.ndim == 3 and frame_1.size(-1) == 3
    N, B, _ = frame_1.shape

    w = torch.as_tensor(w, dtype=frame_1.dtype, device=frame_1.device)
    w = w.expand(N, B)
    w = w.unsqueeze(-1)

    return (1 - w) * frame_1 + w * frame_2
```
